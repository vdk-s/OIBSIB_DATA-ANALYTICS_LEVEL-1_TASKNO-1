```python
plt.figure()
df.groupby('Category')['Calories'].mean().sort_values().plot(kind='barh')
plt.title('Average Calories by Category')
plt.xlabel('Calories')
plt.ylabel('Category')
plt.show()
```

```python
plt.figure()
plt.hist(df['Calories'], bins=30)
plt.title('Calorie Distribution')
plt.xlabel('Calories')
plt.ylabel('Number of Items')
plt.show()
```

```python
plt.figure()
sns.heatmap(df.select_dtypes(include=['int64', 'float64']).corr(), cmap='coolwarm')
plt.title('Correlation Between Nutritional Values')
plt.show()
```
